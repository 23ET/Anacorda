```python
import numpy as np
from IPython.display import Audio
from matplotlib import pyplot as plt
SR = 44100
```

```python
def play(wave, sample_rate = SR):
    return Audio(wave, rate = sample_rate , normalize = False )
```

```python
Hihat_Dur = 1
N_hihat = int(Hihat_Dur * SR)
```

```python
noise_sn = np.random.uniform(-1,1,N_hihat)

def envelope_exp_hihat(env_duration=1, tau=0.02, sample_rate=SR):
    if tau <= 0:
        raise ValueError("tau must be positive.")
    n_env = round(env_duration * sample_rate)
    env = [np.exp(-(i / sample_rate) / tau) for i in range(n_env)]

    return env

env_h = envelope_exp_hihat()

Hihat = env_h * noise_sn
play(Hihat)
```

```python
from scipy.signal import butter, sosfilt

def high_cut(wave, cutoff=500, order=4, sample_rate=SR):

    # Design Butterworth low-pass filter
    sos = butter(
        order,
        cutoff,
        btype="lowpass",
        fs=sample_rate,
        output="sos"
    )

    # Apply filter
    filtered_wave = sosfilt(sos, wave)

    return filtered_wave
```

```python
Hihat_filtered = high_cut(Hihat, cutoff=13000)
play(Hihat_filtered)
```
